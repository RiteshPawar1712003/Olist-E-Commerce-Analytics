```python
invalid_approval=(
    orders["order_approved_at"]<orders["order_purchase_timestamp"]
).sum()

invalid_carrier=(
    orders["order_delivered_carrier_date"]<orders["order_approved_at"]
).sum()

invalid_delivery=(
  orders["order_delivered_customer_date"]<orders["order_delivered_carrier_date"]
).sum()


print("Approval Before purchase",invalid_approval)
print("Carried before Approval",invalid_carrier)
print("customer Delivery before carrier delivery",invalid_delivery)
```

```python
orders["orders_sequence_valid"]=True
orders.loc[
             orders["order_approved_at"].notna() &
        (orders["order_approved_at"]<orders["order_purchase_timestamp"]),
         "orders_sequence_valid"
]=False
```

```python
orders["orders_sequence_valid"].value_counts()
```

```python
orders.loc[
    orders["order_delivered_carrier_date"].notna() &
    orders["order_approved_at"].notna() &
    (orders["order_delivered_carrier_date"] < orders["order_approved_at"]),
    "orders_sequence_valid"
] = False
```

```python
orders["orders_sequence_valid"].value_counts()
```

```python
orders.loc[
    orders["order_delivered_customer_date"].notna() &
    orders["order_delivered_carrier_date"].notna() &
    (orders["order_delivered_customer_date"] < orders["order_delivered_carrier_date"]),
    "orders_sequence_valid"
] = False
```

```python
orders['delivery_time_valid'] = (
    orders['order_delivered_customer_date'].notna() &
    orders['order_delivered_carrier_date'].notna() &
    (
        orders['order_delivered_customer_date'] >=
        orders['order_delivered_carrier_date']
    )
)

orders['delivery_time_valid'].value_counts()
```

```python
orders["Total delevry time"]=(
    orders["order_delivered_customer_date"] - orders['order_delivered_carrier_date']
).dt.total_seconds()/(24*20*60)
```

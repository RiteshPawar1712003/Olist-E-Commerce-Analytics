```python
import pandas as pd
import os


```

```python
data_path = "../data/"

files = os.listdir(data_path)

print("Files in data folder:")

for file in files:
    if file.endswith(".csv"):
        print(file)
```

```python
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

olist_orders_dataset

```python


orders = pd.read_csv(data_path + "olist_orders_dataset.csv")

print(orders.head())                  # Checking first 5 rows
print(orders.shape)                   # Checking rows and columns
print(orders.columns)                 # Checking column names
print(orders.dtypes)                  # Checking data types
print(orders.isnull().sum())          # Checking null values

print(orders["order_status"].value_counts())

print("Duplicate:", orders.duplicated().sum())
print("Unique order id:", orders["order_id"].nunique())
```
